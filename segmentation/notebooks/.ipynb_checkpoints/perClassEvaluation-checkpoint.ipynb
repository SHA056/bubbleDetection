```python
# per-size-class evaluation: pixel recall + object detection counts
import os
import time

import numpy as np
import pandas as pd
import tensorflow as tf
# from pathlib import Path
from astropy.io import fits
from astropy.coordinates import SkyCoord
from astropy.wcs import WCS
from skimage.draw import ellipse
from scipy import ndimage

# phase 0 post-processing: split merged blobs at watershed lines
from skimage.segmentation import watershed
from skimage.feature import peak_local_max

# visual summary: per-class recall, coverage distribution, threshold sensitivity
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# full-mosaic view: validation regions + detection outcomes in galaxy context
from matplotlib.patches import Rectangle

# full-mosaic view on the ORIGINAL (non-normalized) image
from matplotlib.patches import Rectangle
from matplotlib.colors import AsinhNorm

from matplotlib.lines import Line2D
from matplotlib.patches import Patch
```

```python
SEED = 42
PATCH_SIZE = 256
BIN_THRESHOLD = 0.5       # prediction binarization (the defensible default)
COVERAGE_THRESHOLD = 0.5  # bubble counts as detected above this coverage
MIN_BLOB_PX = 20          # fallback floor; overridden below when MIN_BLOB_AUTO
DISTANCE_PC = 9.77e6
N_TRIALS = 200            # random shifts for the chance-coverage control

ACTIVE_STAGE = "stage02"

# Folds to grade. 
#   [None]        -> the frozen split in valRegions.csv
#   [0, 1, 2, 3]  -> the spatial folds from scripts/makeFolds.py --k 4 --rows 2
FOLDS = [0, 1, 2, 3]

# --- phase 0 post-processing (2026-08-07) ---------------------------------
USE_TTA = True        # average the sigmoid map over the 8 dihedral transforms
USE_WATERSHED = True  # split merged blobs before counting them
MIN_BLOB_AUTO = True  # derive the blob-area floor from the classes this stage was trained on

MIN_BLOB_FRACTION = 0.5

RUN_TAG = "postproc_"
# --------------------------------------------------------------------------

FIGURE_DIR = "../outputs/figures"
CLASS_COLORS = {"edgeLarge": "#2a78d6", "giants": "#1baf7a", "large": "#af1b4f", "medium": "#d6892a", "small": "#892ad6", "tiny": "#ff0fe3"}  # CVD-validated pair

FITS_PATH = "../data/raw/fits/jw02107-o039_t018_miri_f770w_i2d.fits"
CATALOGUE_PATH = "../data/processed/metadata/cleanedSegregatedMergedCatalogue.txt"
IMAGE_NORM_PATH = "../data/processed/images/ngc628Norm.npy"
IMAGE_PATH = "../data/processed/images/ngc628.npy"
BOUNDS_PATH = "../data/processed/metadata/curriculum/sizeClassBounds.csv"
MODEL_DIR = "../outputs/models/"
EXPERIMENTS_DIR = "../outputs/predictions"
PLOTS_PATH = "../outputs/figures/plots"

os.makedirs(PLOTS_PATH, exist_ok=True)


def fold_paths(fold):
    if fold is None:
        return {
            "val_regions": "../data/processed/metadata/valRegions.csv",
            "patch_dir": "../data/processed/patches",
            "aug_dir": "../data/processed/augmented",
            "model": os.path.join(MODEL_DIR, f"{ACTIVE_STAGE}bubble_unet.keras"),
            "tag": f"{ACTIVE_STAGE}{RUN_TAG}",
        }
    return {
        "val_regions": f"../data/processed/metadata/folds/fold{fold}/valRegions.csv",
        "patch_dir": f"../data/processed/patches/fold{fold}",
        "aug_dir": f"../data/processed/augmented/fold{fold}",
        "model": os.path.join(MODEL_DIR, f"{ACTIVE_STAGE}fold{fold}bubble_unet.keras"),
        "tag": f"{ACTIVE_STAGE}fold{fold}{RUN_TAG}",
    }


CURRICULUM_STAGES = {
    "baseline": ["tiny", "small", "medium", "large", "giants", "edgeLarge"],
    "stage00": ["edgeLarge"],
    "stage01": ["giants", "edgeLarge"],
    "stage02": ["large", "giants", "edgeLarge"],
    "stage03": ["medium", "large", "giants", "edgeLarge"],
    "stage04": ["small", "medium", "large", "giants", "edgeLarge"],
    "stage05": ["tiny", "small", "medium", "large", "giants", "edgeLarge"],
}

_missing = [f"  fold {f}: {fold_paths(f)[k]}"
            for f in FOLDS for k in ("val_regions", "patch_dir", "aug_dir", "model")
            if not os.path.exists(fold_paths(f)[k])]
if _missing:
    raise FileNotFoundError(
        "Missing inputs for the requested folds:\n" + "\n".join(_missing) +
        "\n\nRun scripts/makeFolds.py --k 4 --rows 2 for the layout, then "
        "modelTraining.ipynb with the same FOLDS and ACTIVE_STAGE."
    )

print(f"ACTIVE_STAGE={ACTIVE_STAGE}  FOLDS={FOLDS}")
for _f in FOLDS:
    print(f"  fold {_f}: tag={fold_paths(_f)['tag']}")
```

```python
# catalogue in pixel coordinates - fold-independent, so loaded once

df_all = pd.read_csv(CATALOGUE_PATH)
header = fits.open(FITS_PATH)[1].header
wcs = WCS(header)

coords = SkyCoord(ra=df_all["RA_DMS"].values, dec=df_all["DEC_DMS"].values, frame="icrs")
df_all["x_pix"], df_all["y_pix"] = wcs.all_world2pix(coords.ra.deg, coords.dec.deg, 0)

pc_per_pixel = abs(header["CDELT1"]) * 3600 * DISTANCE_PC / 206265
df_all["a_pix"] = df_all["SEMI_MAJ_PC"] / pc_per_pixel
df_all["b_pix"] = df_all["SEMI_MIN_PC"] / pc_per_pixel

bounds = pd.read_csv(BOUNDS_PATH)

if "SIZE_CLASS" not in df_all.columns:
    df_all["SIZE_CLASS"] = df_all["AVG_RAD_PC"].apply(
        lambda r: bounds.loc[
            (bounds.lower_bound_pc <= r) & (r < bounds.upper_bound_pc), "SIZE_CLASS"
        ].iloc[0]
    )
    print("SIZE_CLASS recomputed from bounds (catalogue predates catalogueSegregation)")

print("catalogue bubbles:", len(df_all))
print(df_all["SIZE_CLASS"].value_counts().to_string())
```

```python
# phase 0: blob-area floor matched to the stage, and the watershed seed spacing

trained_classes = CURRICULUM_STAGES[ACTIVE_STAGE]
trained = df_all[df_all["SIZE_CLASS"].isin(trained_classes)]
footprint_px = np.pi * trained["a_pix"] * trained["b_pix"]

if MIN_BLOB_AUTO:
    MIN_BLOB_PX = int(MIN_BLOB_FRACTION * footprint_px.min())

# seeds closer together than the smallest trained bubble radius would split a
# single bubble rather than separate two, so that radius sets the spacing
WATERSHED_MIN_DISTANCE = max(2, int(np.sqrt(footprint_px.min() / np.pi)))

print(f"stage {ACTIVE_STAGE} trained on {trained_classes}")
print(f"  smallest catalogue footprint : {footprint_px.min():8.1f} px "
      f"(radius {np.sqrt(footprint_px.min() / np.pi):.2f} px)")
print(f"  MIN_BLOB_PX                  : {MIN_BLOB_PX:8d} px  (was 20)")
print(f"  WATERSHED_MIN_DISTANCE       : {WATERSHED_MIN_DISTANCE:8d} px")
print(f"  USE_TTA={USE_TTA}  USE_WATERSHED={USE_WATERSHED}  RUN_TAG='{RUN_TAG}'")
```

```python
# --- helpers, all fold-independent -------------------------------------------

def predict_tta(model, X, batch_size=4):
    acc = None
    for k in range(4):
        for flip in (False, True):
            Xt = np.rot90(X, k, axes=(1, 2))
            if flip:
                Xt = Xt[:, :, ::-1, :]
            p = np.squeeze(model.predict(Xt, batch_size=batch_size, verbose=0))
            # undo in reverse order: flip first, then the rotation
            if flip:
                p = p[:, :, ::-1]
            p = np.rot90(p, -k, axes=(1, 2))
            acc = p if acc is None else acc + p
    return acc / 8.0


def label_blobs(mask, use_watershed=None):
    if use_watershed is None:
        use_watershed = USE_WATERSHED
    if not use_watershed:
        return ndimage.label(mask)
    distance = ndimage.distance_transform_edt(mask)
    coords = peak_local_max(distance, labels=mask.astype(int),
                            min_distance=WATERSHED_MIN_DISTANCE,
                            exclude_border=False)
    if len(coords) == 0:
        return ndimage.label(mask)
    seeds = np.zeros(mask.shape, dtype=bool)
    seeds[tuple(coords.T)] = True
    markers, _ = ndimage.label(seeds)
    labels = watershed(-distance, markers, mask=mask)
    return labels, int(labels.max())


def graded_catalogue(val_regions):
    def centre_in_val_region(y, x):
        for _, region in val_regions.iterrows():
            if (region.y_start <= y < region.y_start + region.height and
                    region.x_start <= x < region.x_start + region.width):
                return True
        return False

    keep = [centre_in_val_region(b.y_pix, b.x_pix) for _, b in df_all.iterrows()]
    return df_all[keep].reset_index(drop=True)


def render_footprints(val_meta, graded_ids, Y_val_ref):
    active = set(CURRICULUM_STAGES[ACTIVE_STAGE])

    records = []
    union_all = np.zeros_like(np.squeeze(Y_val_ref), dtype=bool)
    union_stage = np.zeros_like(union_all)

    for p, row in val_meta.iterrows():
        ys, xs = int(row.y_start), int(row.x_start)
        for _, bub in df_all.iterrows():
            r_max = max(bub.a_pix, bub.b_pix)
            if (bub.y_pix + r_max < ys or bub.y_pix - r_max >= ys + PATCH_SIZE or
                    bub.x_pix + r_max < xs or bub.x_pix - r_max >= xs + PATCH_SIZE):
                continue
            rr, cc = ellipse(
                r=bub.y_pix - ys, c=bub.x_pix - xs,
                r_radius=bub.b_pix, c_radius=bub.a_pix,
                rotation=np.deg2rad(bub.PA_DEG), shape=(PATCH_SIZE, PATCH_SIZE),
            )
            if len(rr) == 0:
                continue
            union_all[p][rr, cc] = True
            if bub.SIZE_CLASS in active:
                union_stage[p][rr, cc] = True
            if bub.GLOBAL_ID in graded_ids:   # only exam bubbles are graded for recall
                records.append(dict(patch=p, global_id=bub.GLOBAL_ID,
                                    size_class=bub.SIZE_CLASS, rr=rr, cc=cc))

    agreement = (union_stage == (np.squeeze(Y_val_ref) > 0.5)).mean()
    return records, union_all, union_stage, agreement


def summarise(bubbles):
    """Per-class object/pixel recall from a per-bubble coverage table."""
    return bubbles.groupby("size_class").apply(lambda g: pd.Series({
        "n_bubbles": len(g),
        "n_detected": int(g.detected.sum()),
        "object_recall": g.detected.mean(),
        "pixel_recall": g.visible_px.mul(g.coverage).sum() / g.visible_px.sum(),
        "mean_coverage": g.coverage.mean(),
    }))
```

```python
def evaluate_fold(fold):
    paths = fold_paths(fold)
    tag = paths["tag"]
    print(f"\n{'=' * 70}\nFOLD {fold}   tag={tag}\n{'=' * 70}")

    # --- validation set from the split this fold's patches recorded ---
    X = np.load(os.path.join(paths["patch_dir"], "X.npy"))
    Y = np.load(os.path.join(paths["patch_dir"], "Y.npy"))
    metadata = pd.read_csv(os.path.join(paths["patch_dir"], "patchMetadata.csv"))
    X_val_ref = np.load(os.path.join(paths["aug_dir"], "XVal.npy"))
    Y_val_ref = np.load(os.path.join(paths["aug_dir"], "YVal.npy"))

    assert "split" in metadata.columns, (
        "patchMetadata.csv has no 'split' column - re-run modelTraining for this fold"
    )
    val_idx = metadata.index[metadata["split"] == "val"].to_numpy()
    assert np.array_equal(np.squeeze(X[val_idx]), np.squeeze(X_val_ref)), \
        f"fold {fold}: split mismatch - patch and augmented arrays disagree"
    assert np.array_equal(np.squeeze(Y[val_idx]), np.squeeze(Y_val_ref)), \
        f"fold {fold}: split mismatch - patch and augmented arrays disagree"

    val_meta = metadata.iloc[val_idx].reset_index(drop=True)

    # --- which bubbles this fold is allowed to grade ---
    val_regions = pd.read_csv(paths["val_regions"])
    df = graded_catalogue(val_regions)
    graded_ids = set(df["GLOBAL_ID"])

    print(f"val patches {len(val_meta)} | graded bubbles {len(df)}")
    print(df["SIZE_CLASS"].value_counts().to_string())

    records, union, union_stage, agreement = render_footprints(val_meta, graded_ids, Y_val_ref)
    print(f"footprint union vs Y_val agreement: {agreement:.6f}")
    assert agreement > 0.999, (
        f"fold {fold}: stage-filtered rendering does not reproduce Y_val "
        f"(agreement {agreement:.6f}). The mask on disk and ACTIVE_STAGE="
        f"{ACTIVE_STAGE} disagree - re-run galaxyMaskGeneration for this stage."
    )

    # --- predict ---
    model = tf.keras.models.load_model(paths["model"], compile=False)
    pred_soft = (predict_tta(model, X_val_ref) if USE_TTA
                 else np.squeeze(model.predict(X_val_ref, batch_size=4)))
    pred_bin = pred_soft > BIN_THRESHOLD
    print(f"painted fraction at t={BIN_THRESHOLD}: {pred_bin.mean():.4f}"
          f"   (TTA {'on' if USE_TTA else 'off'})")

    # --- per-bubble coverage ---
    per_bubble = {}
    for rec in records:
        covered = int(pred_bin[rec["patch"]][rec["rr"], rec["cc"]].sum())
        e = per_bubble.setdefault(rec["global_id"],
            dict(size_class=rec["size_class"], visible=0, covered=0))
        e["visible"] += len(rec["rr"])
        e["covered"] += covered

    bubbles = pd.DataFrame([
        dict(global_id=k, size_class=v["size_class"],
             visible_px=v["visible"], coverage=v["covered"] / v["visible"])
        for k, v in per_bubble.items()
    ])
    bubbles["detected"] = bubbles["coverage"] >= COVERAGE_THRESHOLD
    bubbles["fold"] = -1 if fold is None else fold

    summary = summarise(bubbles)

    # --- blob precision ---
    n_blobs = n_matched = 0
    for p in range(len(pred_bin)):
        labels, n = label_blobs(pred_bin[p])
        for i in range(1, n + 1):
            blob = labels == i
            if blob.sum() < MIN_BLOB_PX:
                continue
            n_blobs += 1
            n_matched += bool((blob & union[p]).any())

    precision = n_matched / n_blobs if n_blobs else float("nan")
    print(f"predicted blobs: {n_blobs} | matched: {n_matched} | "
          f"spurious: {n_blobs - n_matched} | precision: {precision:.3f}")

    summary["global_precision"] = precision
    summary["global_n_pred_blobs"] = n_blobs
    summary["global_n_spurious"] = n_blobs - n_matched
    summary["pp_min_blob_px"] = MIN_BLOB_PX
    summary["pp_watershed"] = USE_WATERSHED
    summary["pp_tta"] = USE_TTA
    summary["pp_painted_fraction"] = float(pred_bin.mean())

    # --- chance-coverage control, on this fold's own predictions ---
    rng = np.random.default_rng(SEED)
    bubble_meta = {}
    for rec in records:
        e = bubble_meta.setdefault(rec["global_id"],
            dict(size_class=rec["size_class"], visible=0))
        e["visible"] += len(rec["rr"])

    class_ids = {cls: [g for g, m in bubble_meta.items() if m["size_class"] == cls]
                 for cls in CURRICULUM_STAGES[ACTIVE_STAGE]}
    trial_recalls = {cls: [] for cls in class_ids}
    # per-bubble detection rate under the null, kept so folds can be pooled
    null_hits = {gid: 0 for gid in bubble_meta}

    for _ in range(N_TRIALS):
        shifted = [np.roll(pred_bin[p_i],
                           shift=(rng.integers(PATCH_SIZE), rng.integers(PATCH_SIZE)),
                           axis=(0, 1))
                   for p_i in range(len(pred_bin))]
        covered = {gid: 0 for gid in bubble_meta}
        for rec in records:
            covered[rec["global_id"]] += int(shifted[rec["patch"]][rec["rr"], rec["cc"]].sum())
        detected = {gid: covered[gid] / m["visible"] >= COVERAGE_THRESHOLD
                    for gid, m in bubble_meta.items()}
        for gid, hit in detected.items():
            null_hits[gid] += int(hit)
        for cls, ids in class_ids.items():
            if ids:
                trial_recalls[cls].append(np.mean([detected[g] for g in ids]))

    control = pd.DataFrame([
        dict(size_class=cls,
             n_bubbles=len(class_ids[cls]),
             observed_recall=summary.loc[cls, "object_recall"],
             chance_recall_mean=np.mean(trial_recalls[cls]),
             chance_recall_std=np.std(trial_recalls[cls]),
             chance_recall_p95=np.percentile(trial_recalls[cls], 95),
             excess_above_chance=summary.loc[cls, "object_recall"] - np.mean(trial_recalls[cls]))
        for cls in class_ids if class_ids[cls]
    ]).set_index("size_class")
    control["pp_tta"] = USE_TTA
    control["pp_painted_fraction"] = float(pred_bin.mean())

    bubbles["null_rate"] = bubbles["global_id"].map(
        {g: h / N_TRIALS for g, h in null_hits.items()}
    )

    # --- save per fold ---
    summary.to_csv(os.path.join(EXPERIMENTS_DIR, f"{tag}perClassEval.csv"))
    bubbles.to_csv(os.path.join(EXPERIMENTS_DIR, f"{tag}perBubbleCoverage.csv"), index=False)
    control.to_csv(os.path.join(EXPERIMENTS_DIR, f"{tag}chanceCoverageControl.csv"))

    return dict(
        fold=fold, tag=tag, paths=paths,
        val_meta=val_meta, df=df, records=records,
        union=union, union_stage=union_stage,
        X_val_ref=X_val_ref, Y_val_ref=Y_val_ref,
        pred_soft=pred_soft, pred_bin=pred_bin,
        bubbles=bubbles, summary=summary, control=control,
        n_blobs=n_blobs, n_matched=n_matched, precision=precision,
    )
```

```python
results = {fold: evaluate_fold(fold) for fold in FOLDS}

print("\n" + "=" * 70)
print(f"{'fold':<8}{'val pt':>8}{'graded':>8}{'blobs':>8}{'spur':>7}{'prec':>8}{'painted':>9}")
for fold, r in results.items():
    print(f"{str(fold):<8}{len(r['val_meta']):>8}{len(r['bubbles']):>8}"
          f"{r['n_blobs']:>8}{r['n_blobs'] - r['n_matched']:>7}"
          f"{r['precision']:>8.3f}{r['pred_bin'].mean():>9.4f}")
```

```python
# --- aggregation 1: per-fold mean and spread ---------------------------------

per_fold = pd.concat(
    [r["summary"].assign(fold=(-1 if f is None else f)).reset_index()
     for f, r in results.items()],
    ignore_index=True,
)
per_fold.to_csv(os.path.join(EXPERIMENTS_DIR,
                             f"{ACTIVE_STAGE}{RUN_TAG}perFoldPerClass.csv"), index=False)

if len(results) > 1:
    spread = per_fold.groupby("size_class").agg(
        n_folds=("fold", "nunique"),
        n_bubbles=("n_bubbles", "sum"),
        recall_mean=("object_recall", "mean"),
        recall_std=("object_recall", "std"),
        recall_min=("object_recall", "min"),
        recall_max=("object_recall", "max"),
        precision_mean=("global_precision", "mean"),
    )
    print(f"per-fold spread across {len(results)} folds:\n")
    print(spread.to_string(float_format="%.4f"))
else:
    print("single fold - set FOLDS = [0, 1, 2, 3] for a spread")


# --- aggregation 2: pooled per-bubble, the cross-validated result ------------

pooled_bubbles = pd.concat([r["bubbles"] for r in results.values()], ignore_index=True)

dupes = pooled_bubbles["global_id"].duplicated().sum()
if dupes:
    print(f"\nWARNING: {dupes} bubbles graded by more than one fold. The fold "
          "blocks are not a clean partition, so this pooled table double-counts "
          "them - check the makeFolds layout before quoting it.")

pooled = summarise(pooled_bubbles)

# chance floor pooled from each bubble's own null detection rate, so a bubble
# keeps the floor of the fold that actually graded it
pooled["chance_recall_mean"] = pooled_bubbles.groupby("size_class")["null_rate"].mean()
pooled["excess_above_chance"] = pooled["object_recall"] - pooled["chance_recall_mean"]

pooled.to_csv(os.path.join(EXPERIMENTS_DIR,
                           f"{ACTIVE_STAGE}{RUN_TAG}pooledPerClassEval.csv"))
pooled_bubbles.to_csv(os.path.join(EXPERIMENTS_DIR,
                                   f"{ACTIVE_STAGE}{RUN_TAG}pooledPerBubbleCoverage.csv"),
                      index=False)

print(f"\npooled over {len(results)} fold(s): {len(pooled_bubbles)} bubbles graded "
      f"of {len(df_all)} in the catalogue\n")
print(pooled[["n_bubbles", "n_detected", "object_recall",
              "chance_recall_mean", "excess_above_chance"]].to_string(float_format="%.4f"))
```

```python
# --- operating-point sweep: recall and precision vs binarization threshold -----

SWEEP_GRID = np.round(np.arange(0.05, 0.96, 0.05), 3)

SWEEP_CHANCE_TRIALS = 50


def sweep_point(r, t, n_trials=SWEEP_CHANCE_TRIALS):
    pred_bin = r["pred_soft"] > t
    records, union = r["records"], r["union"]

    # --- per-bubble coverage ---
    per_bubble = {}
    for rec in records:
        covered = int(pred_bin[rec["patch"]][rec["rr"], rec["cc"]].sum())
        e = per_bubble.setdefault(rec["global_id"],
            dict(size_class=rec["size_class"], visible=0, covered=0))
        e["visible"] += len(rec["rr"])
        e["covered"] += covered

    bubbles = pd.DataFrame([
        dict(global_id=k, size_class=v["size_class"], visible_px=v["visible"],
             coverage=v["covered"] / v["visible"])
        for k, v in per_bubble.items()
    ])
    bubbles["detected"] = bubbles["coverage"] >= COVERAGE_THRESHOLD

    per_class = bubbles.groupby("size_class")["detected"].agg(["size", "sum", "mean"])
    per_class.columns = ["n_bubbles", "n_detected", "object_recall"]

    # --- blob precision, and the blob size distribution behind it ---
    n_blobs = n_matched = 0
    blob_px = []
    for p in range(len(pred_bin)):
        labels, n = label_blobs(pred_bin[p])
        for i in range(1, n + 1):
            blob = labels == i
            size = int(blob.sum())
            if size < MIN_BLOB_PX:
                continue
            n_blobs += 1
            blob_px.append(size)
            n_matched += bool((blob & union[p]).any())

    # --- chance floor on THIS binarization ---
    if n_trials:
        rng = np.random.default_rng(SEED)
        meta = {}
        for rec in records:
            e = meta.setdefault(rec["global_id"],
                                dict(size_class=rec["size_class"], visible=0))
            e["visible"] += len(rec["rr"])
        class_ids = {c: [g for g, m in meta.items() if m["size_class"] == c]
                     for c in per_class.index}
        trials = {c: [] for c in class_ids}
        for _ in range(n_trials):
            shifted = [np.roll(pred_bin[p_i],
                               shift=(rng.integers(PATCH_SIZE), rng.integers(PATCH_SIZE)),
                               axis=(0, 1))
                       for p_i in range(len(pred_bin))]
            covered = {g: 0 for g in meta}
            for rec in records:
                covered[rec["global_id"]] += int(
                    shifted[rec["patch"]][rec["rr"], rec["cc"]].sum())
            det = {g: covered[g] / m["visible"] >= COVERAGE_THRESHOLD
                   for g, m in meta.items()}
            for c, ids in class_ids.items():
                if ids:
                    trials[c].append(np.mean([det[g] for g in ids]))
        per_class["chance_p95"] = [np.percentile(trials[c], 95) for c in per_class.index]
    else:
        per_class["chance_p95"] = np.nan

    per_class["threshold"] = t
    per_class["fold"] = -1 if r["fold"] is None else r["fold"]
    per_class["precision"] = n_matched / n_blobs if n_blobs else np.nan
    per_class["n_pred_blobs"] = n_blobs
    per_class["n_spurious"] = n_blobs - n_matched
    per_class["median_blob_px"] = float(np.median(blob_px)) if blob_px else np.nan
    per_class["painted_fraction"] = float(pred_bin.mean())
    return per_class.reset_index()


_rows = []
for fold, r in results.items():
    t0 = time.time()
    for t in SWEEP_GRID:
        _rows.append(sweep_point(r, t))
    print(f"fold {fold}: {len(SWEEP_GRID)} thresholds in {time.time() - t0:.0f}s")

sweep = pd.concat(_rows, ignore_index=True)

# the sweep must reproduce the headline evaluation at the operating threshold,
# or it is measuring something else
_check = sweep[np.isclose(sweep["threshold"], BIN_THRESHOLD)]
if len(_check):
    for fold, r in results.items():
        f = -1 if fold is None else fold
        got = _check[_check["fold"] == f]["painted_fraction"].iloc[0]
        assert np.isclose(got, r["pred_bin"].mean()), (
            f"fold {fold}: sweep paints {got:.4f} at t={BIN_THRESHOLD} but "
            f"evaluate_fold painted {r['pred_bin'].mean():.4f}"
        )
    print(f"sweep reproduces the headline painted fraction at t={BIN_THRESHOLD}")
else:
    print(f"note: BIN_THRESHOLD={BIN_THRESHOLD} is not on SWEEP_GRID, "
          "so the sweep cannot be cross-checked against the headline numbers")

sweep_path = os.path.join(EXPERIMENTS_DIR, f"{ACTIVE_STAGE}{RUN_TAG}thresholdSweep.csv")
sweep.to_csv(sweep_path, index=False)
print("wrote", sweep_path)
```

```python
# watershed diagnostics: is it separating merged bubbles, or shredding single ones?

records_by_patch = {}
for rec in records:
    records_by_patch.setdefault(rec["patch"], []).append(rec)


def watershed_diagnostics(use_watershed):
    n_blobs = 0
    dominant_of = {}   # (patch, blob_id) -> list of graded bubble ids it owns
    split_bubbles = 0
    for p in range(len(pred_bin)):
        labels, n = label_blobs(pred_bin[p], use_watershed=use_watershed)
        counts = np.bincount(labels.ravel(), minlength=n + 1)
        kept = {i for i in range(1, n + 1) if counts[i] >= MIN_BLOB_PX}
        n_blobs += len(kept)
        for rec in records_by_patch.get(p, []):
            lab = labels[rec["rr"], rec["cc"]]
            lab = lab[np.isin(lab, list(kept))] if kept else lab[:0]
            if lab.size == 0:
                continue
            hit = np.bincount(lab)
            # a bubble counts as split when a second blob covers a real share of
            # it, not just a few boundary pixels
            share = hit / hit.sum()
            if (share >= 0.15).sum() >= 2:
                split_bubbles += 1
            dominant_of.setdefault((p, int(hit.argmax())), []).append(rec["global_id"])
    merged_blobs = sum(1 for v in dominant_of.values() if len(set(v)) >= 2)
    return n_blobs, merged_blobs, split_bubbles


n_graded = len({r["global_id"] for r in records})
print(f"graded bubbles in the val patches: {n_graded}\n")
print(f"{'':<16}{'blobs':>8}{'merged':>9}{'split':>8}")
for flag, name in [(False, "components"), (True, "watershed")]:
    nb, mb, sb = watershed_diagnostics(flag)
    print(f"{name:<16}{nb:>8}{mb:>9}{sb:>8}")
print("\nwant: blobs up toward the graded count, merged down, split roughly flat")
```
